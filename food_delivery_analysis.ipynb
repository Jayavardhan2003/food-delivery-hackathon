```python
def rating_bucket(r):
    if 3.0 <= r <= 3.5: return '3.0-3.5'
    if 3.6 <= r <= 4.0: return '3.6-4.0'
    if 4.1 <= r <= 4.5: return '4.1-4.5'
    return'4.6-5.0'
df['rating_range'] = df['rating'].apply(rating_bucket)

df.groupby('rating_range')['total_amount'].sum().sort_values(ascending=False)
```
